# Missing a Single Number

```python
array = [5, 6, 7, 8, 9, 11, 12]
```

```python
def search_missing(array):
    
    arr = sorted(array)

    for i in range(len(arr) - 1):
        if (arr[i + 1] != arr[i] + 1):
            return f"Missing element: {arr[i] + 1}"
        
    return f"No element is missing."
```

```python
print(search_missing(array))
```

# Missing Numbers in a Sequence

```python
arr = [5, 6, 7, 8, 12, 14, 20]
```

```python
def search_missing_elements(array):
    
    arr = sorted(array)

    missing = []
    
    for i in range(len(arr) - 1):
        if (arr[i + 1] - arr[i]) > 1:
            missing.extend(range(arr[i] + 1, arr[i + 1]))
    
    return missing
```

```python
print(f"All missing elements: {search_missing_elements(arr)}")
```
