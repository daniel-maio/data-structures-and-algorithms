# Longest Name in List

```python
def longest_name(names: list[str]):
    
    l_ind = []
    
    for i in range(0, len(names)):
        l_ind.append(len(names[i]))

    idx = l_ind.index(max(l_ind))

    return f"The longest name is: {names[idx]}, at index {idx}, with {max(l_ind)} chars."
```

```python
names = ["John", "Herbert", "Urnold_Schwarzenege", "Cryslaine", "Arnold_Schwarzeneger", "Tim"]
```

```python
print(longest_name(names))
```
