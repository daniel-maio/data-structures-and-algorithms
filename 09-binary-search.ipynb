# Binary Search

Binary Search is a searching algorithm that operates on a **sorted** or monotonic search space, repeatedly dividing it into halves to find a target value or optimal answer in logarithmic time O(log N).

```python
array = [0, 1, 5, 10, 13, 25, 85, 92, 93, 99]
len(array)
```

```python
def binary_search(arr, target):

    n = len(arr)
    
    mid = n // 2

    if arr[mid] == target:
        return f'Element {target} is in array.'
    
    if mid != 0:
        if arr[mid] > target:
            return binary_search(arr[:mid], target)
        elif arr[mid] < target:
            return binary_search(arr[mid:], target)
    
    else:
        print(f'Element {target} is NOT in array.')
```

```python
target = 94
```

```python
binary_search(array, target)
```
