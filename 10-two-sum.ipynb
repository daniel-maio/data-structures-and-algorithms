# Two Sum

Given an array of n integers and a target value, check if there exists a pair whose sum equals the target.\
This is a variation of the 2-Sum problem.

Input: arr = [0, -1, 2, -3, 1], target = -2\
-> There is a pair (1, -3) with the sum equal to given target, 1 + (-3) = -2.

```python
def two_sum(arr, target):

    n = len(arr)
    
    dict = {}
    
    for i in range(n):
        
        diff = target - arr[i]
        
        if diff in arr:
            dict.update({tuple(sorted((diff, arr[i]))): diff+arr[i]})
        else:
            pass
    
    return dict
```

```python
array = [1, 5, 10, -5, 22, 20, 15, 2, 12, 6, 30]
```

```python
target = 25
```

```python
two_sum(array, target)
```
