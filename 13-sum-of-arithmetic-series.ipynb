# Sum of Arithmetic Series

A series with same common difference is known as arithmetic series. The first term of series is 'a' and common difference is d.

The series looks like a, a + d, a + 2d, a + 3d, . . .

Find the sum of series up to nth term.

Example:

Input: n = 5, a = 1, d = 3

s = [1, 4, 7, 10, 13], so sum will be 35.

```python
def sum_of_series(n, a, d):
    
    l = []

    total = 0

    for i in range(n):
        s = a + i * d
        l.append(s)
        total += s
    
    return l, total
```

```python
# change the parameters for different results

n = 5
a = 1
d = 3

arr, arr_sum = sum_of_series(n,a,d)
```

```python
print(f"Array: {arr} | Sum of elements in array = {arr_sum}")
```
