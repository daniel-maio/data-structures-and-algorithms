# Power of 2

Given a positive integer n, check whether it is a power of 2 or not.

Input : n = 16\
Output :  true\
Explanation: 2^4 = 16

```python
def check_power_of_2(value: int):
    
    if value == 0:
        return False
        
    count = 0
    
    while value > 0:
        count += (value & 1) # bitwise operation: compare binary n with binary 1.
        value >>= 1 # right shift n to remove the rightmost bit.
    
    return count == 1
```

```python
value = 512
```

```python
print(check_power_of_2(value))
```
