# Palindrome Array

Given an array, the task is to determine whether an array is a palindrome or not.

Example:

arr = [1, 2, 3, 2, 1]

Output: True

```python
def check_palindrome(arr):
    
    n = len(arr)

    for i in range(n // 2):
        
        if arr[i] != arr[n - 1 - i]:
            return False
    
    return True
```

```python
array = [1, 2, 2, 1]
#array = [1, 2, 9, 1, 2]
#array = [1, 2, 3, -4, 11, -4, 3, 2, 1]

#array = "abcdedcba"
# array = "adcdeabcd"
#array = "xyzxyz"
```

```python
check_palindrome(array)
```
