# Linear Search

```python
def linear_search(arr, key):
    for i in arr:
        if key == i:
            return f'Element {key} is in array at index: {arr.index(key)}.'
        else:
            pass
    return f'Element {key} is not in array.'
```

```python
array = [8, 13, 2, 56, 90, 115, 3, 7, 33]
```

```python
key = 9
```

```python
print(linear_search(array, key))
```
